```python
# p40 처음 만드는 머신러닝 모델

# 실습에 필요한 라이브러리를 불러옵니다.
from sklearn import datasets
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# 사이킷런에서 붓꽃 품종 데이터세트를 가져옵니다.
dataset = datasets.load_iris()

# 입력 데이터와 타깃을 준비합니다.
X, y = dataset['data'], dataset['target']

# K-최근접 이웃 모델 객체를 만듭니다.
model = KNeighborsClassifier(n_neighbors=5)

# K-최근접 이웃 모델에 입력 데이터와 타깃을 넣고 학습시킵니다.
model.fit(X, y)

# 학습된 모델에 입력 데이터를 다시 넣고 타깃을 추론합니다.
y_predicted = model.predict(X)

# 모델이 추론한 타깃과 실제 타깃을 화면에 출력합니다.
print('모델이 추론한 타깃: {labels}'.format(labels=y_predicted))
print('실제 타깃: {labels}'.format(labels=y))

# 모델이 추론한 타깃과 실제 타깃으로 정확도를 계산해 출력합니다.
score = np.mean(y_predicted == y)
print('모델 정확도: {:.3f}'.format(score))

print(y_predicted == y)
print("틀린 개수: ", sum(y_predicted != y))
```

```python
# p45 프로젝트 개선하기 


from sklearn import datasets
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split

# 사이킷런에서 붓꽃 품종 데이터세트를 가져옵니다.
dataset = datasets.load_iris()
print('\n data dataset \n : ', dataset['data'])
print('\n target dataset \n : ', dataset['target'])

# 입력 데이터와 타깃을 준비합니다.
X, y = dataset['data'], dataset['target']

# 데이터세트를 학습 세트와 테스트 세트로 분할합니다.
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)


# K-최근접 이웃 모델 객체를 만듭니다.
model = KNeighborsClassifier(n_neighbors=5)

# K-최근접 이웃 모델에 학습 세트의 입력 데이터와 타깃을 넣고 학습시킵니다.
model.fit(X_train, y_train)


# 학습 세트에서의 모델 정확도와 테스트 세트에서의 모델 정확도를 계산합니다.
train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

# 학습 세트에서의 정확도와 테스트 세트에서의 정확도를 출력합니다.
print('학습 세트 정확도: {:.3f}'.format(train_score))
print('테스트 세트 정확도: {:.3f}'.format(test_score))
```

```python
from sklearn import datasets
import pandas as pd
import matplotlib.pyplot as plt


# 붓꽃 품종 데이터세트를 불러옵니다.
dataset = datasets.load_iris()

# 입력 데이터와 타깃을 준비합니다.
X, y = dataset['data'], dataset['target']


# 넘파이 배열 형태인 기존의 데이터를 판다스의 데이터 프레임으로 변환합니다.
column_names = ['sepal length', 'sepal width', 'petal length', 'petal width']
df = pd.DataFrame(X, columns=column_names)


# 판다스의 기능을 이용해 산점도 행렬을 추가합니다.
pd.plotting.scatter_matrix(df, c=y)

# 그래프를 화면에 그립니다. 
plt.show()
```
