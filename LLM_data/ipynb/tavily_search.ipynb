2025년 경제성장률, 1.0%p 하향조정한 0.7% ￭ 한국 경제 수정 전망의 배경 현대경제연구원은 '25년 한국 경제성장률에 대하여 지난 전망치 1.7%('24년 12월 전망 시점)에서 0.7%로 1.0%p 하향조정하였다. 하향조정의 가장 큰 이유는 소비와 투자의 내수 불황이 장기화되는 가운데, 경기 선행지표들의 뚜렷한 반등 신호를 포 착할 수 없기 때문이다. 또한, 향후 대외 불확실성(트럼프 관세 인상 정책의 파급 영향)의 부정적 영향이 나타날 것이기 때문에, 이제부터 수출 경기 침체가 본격화될 가능성이 높다고 분석된다. 나아가 '24년 2분기부터 시작된 불황이 만 1년 동안이나 지속 중인 상황임에도, 경기 부양을 위한 적극적인 정책적 대응이 보이지 않는다는 점까지 고려할 때, 한국 경제에 대해 긍정적 시각을 가지기 어렵다고 판단된다. 향후 한국 경제는 수출 경기와 내수 경기가 동반 침체되는 ‘절대 수요 부족’ 국 면에 직면할 것이며, 이에 따라 「기업 실적 악화 → 고용 시장 냉각 → 소비 침체 → 시장 수요 위축 → 기업 실적 악화」의 경기 악순환에 빠질 가능성이 우려된다. 내수 부문은 고금리에 따르는 소비자 실질구매력의 정체와 성장과 고용에 대한 국 민경제적 영향이 매우 높은 건설투자의 침체로, 민간 부문 스스로의 회복력이 거의 고갈된 상황이다. 과거의 경험상 한국 경제가 불황 국면에서 탈출하는 전형적인 경 로는 수출 경기 호조가 내수 회복으로 이어지는 것이나, 올해 교역 환경의 급격한 악화로 수출의 경제 성장 견인력을 기대하기 어렵다고 판단된다. 따라서 기본 시나리오 하에서는 '25년 한국 경제의 0%대의 초 저성장 국면 탈출은 불가능해 보이며, 긍정적 시나리오상 ① 교역 환경의 주목할만한 개선이 이루어지 거나 ② 적극적인 경기부양책이 대응될 경우 1%대 초반의 성장률을 달성할 여지는 있다고 판단된다.\n￭ 한국 경제 수정 전망 ① 경제 성장 GDP증가율 2.0% ⇨ 0.7% (2024) (2025) 한국 경제는 '24년 내수 침체 속 수출 호조 국면에서 '25년에는 내수와 수출이 모 EXECUTIVE SUMMARY ii 두 침체되는 국면으로의 진입이 전망된다. (적극적인 재정정책 추진) 대내외 불확실성이 그 어느 때보다 큰 상황인 만 큼 민간 중심의 성장경로 재진입 여부가 확인될 때까지 적극적인 재정정책과 성장 친화적 경제정책 기조를 유지해야 함 ․ 국내 실물경기는 내·외수 동반 부진으로 경제 전반의 수요 압력이 약화 되는 전형적인 침체 국면이 지속될 것으로 판단됨 ․ 더구나, 韓美 간 관세협상에서 긍정적인 성과를 도출하더라도 미국의 對 세계 관세정책의 불확실성과 불투명성이 크다는 점을 고려하면 외수 회복 시기도 불투명함 ․ 따라서, 적극적인 재정정책과 더불어 성장 친화적 경제정책 기조 유지로 내수 경기 회복에 집중하는 한편, 상황 악화 시 연내 추가적인 재정 지출 확대 조치도 취할 수 있을 정도의 유연성도 갖추어야 하는 상황임 ․ 한편, 재정 여력 및 효율성을 고려할 때 경제적 파급 효과가 큰 SOC 분야 에 대한 사업 조기 발주와 함께 가계 구매력 확충과 직결된 부문의 재정 집행 속도를 높여 내수 전반의 경기 침체 방어에 주력해야 함 ․ 재정 건전성 측면에서는 경제가 위중한 상황이라는 점을 고려해 단기 세 수·세출 균형을 고집할 것이 아니라, 중기 세수·세출 균형을 위한 경제 및 재정 전략을 마련하고 실천해 나가는 것이 바람직함 - (추가적 통화정책 완화) 통화정책은 재정정책과의 정합성 확보로 경기부양효 과의 극대화를 꾀하는 한편 물가 불안 등의 우려는 미시 대응을 통해 보완할 필요가 있음 ․ 국내 경제의 수요 압력 약화로 인플레 우려는 상대적으로 낮은 것으로 판 단됨에 따라 추가적 통화 완화가 물가 부담으로 직결되지는 않을 전망임 ․ 따라서, 일시적 계절적 물가 불안 요인에 대해서는 사전에 적절한 수급 대 책을 확보해 즉시 대응이 가능하도록 하는 한편 공공요금 인상 등 정책적 물가 불안 요인은 시기와 범위 조정 등 관리 강화가 필요함 2025년 한국 경제 전망(수정) 27 ․ 또, 추가적인 통화 완화에 있어서 걸림돌로 작용하고 있는 부동산 시장과 가계부채 문제는 정책 일관성 확보를 통한 시장 기대심리 안정화, 가계대 출에 대한 총량 규제 활용 및 고정금리 대출 비중 확대 등과 같은 미시 정책 대응을 통해 보완할 필요가 있음 - (금융시장의 안정성 확보) 금융시장 급변동에 의한 경제주체의 심리 약화가 내수 부진을 가속화하지 않도록 가격 변수의 변동성 완화와 금융기관의 건 전성 제고를 위한 노력도 지속되어야 함 ․ 환율, 주가, 금리 등 금융시장 가격 변수들은 실물 경제적 측면에서 당장 큰 영향을 미치지 않지만, 심리적 측면에서 가계와 기업의 경제 활력에 우 회적인 영향을 미친다는 점을 고려할 필요가 있음 ․ 외환 및 주식 시장 등 금융시장 불안은 경제주체의 심리 악화를 통해 소 비나 투자에 악영향을 미쳐 내수 경기의 회복을 제약할 수 있음 ․ 따라서 금융시장 안전성 확보를 위한 자본 유출입 모니터링 강화, 일부 투 기 자본의 교란 행위 방지를 위한 시스템 보완 등의 노력이 필요함 ․ 특히, 글로벌 자산 시장 변동성 확대가 국내 금융기관이나 개인의 피해로 이어지지 않도록 재무 및 금융 자산 투자 건전성 감독에 주력해야 함 ․ 한편, 금융시장 안정을 위해서는 정책당국과 시장 간 의사소통의 역할도 매우 중요한 만큼 ‘포워드 가이던스(Foward Guidance)’와 실제 정책의사 결정 결과와의 간극을 좁힘으로써 정책당국에 대한 시장의 신뢰와 건전한 기대를 형성해 급변동을 예방하는 것이 중요함 둘째, 트럼프 2기 행정부의 관세정책 변화 양상 및 주요국과의 협상 프로세스 등에 대한 모니터링 강화를 통해 국익 최대화를 위한 협상 전략을 수립하 는 한편, 통상전략 전반의 재점검 및 보완책 마련 등을 통해 피해 최소화 및 수출 경쟁력 유지를 꾀해야 한다. - (주요국 협상 모니터링을 통한 보완책 모색) 트럼프 2기 행정부의 관세협상 이 관세뿐 아니라 안보(국방, 영토), 통화(환율), 첨단 기술 및 산업(반도체, 자동 차, 에너지 등) 등과 연계되어 있다는 점을 고려할 때 주요국과의 협상 과정에 대한 모니터링 결과를 협상전략에 반영할 수 있도록 해야 함 2025년 한국 경제 전망(수정) 28 ․ 트럼프 2기 행정부는 우리나라를 포함한 영국과 호주, 인도, 일본 등 5개 국을 최우선 협상국으로 선정, ‘원칙적 합의’와 ‘구체화 실무작업’의 2단계로 무역 협상을 진행 중임 ․ 이 가운데 일본의 경우, 주일미군주둔비와 환율 등이 비관세장벽과 함께 관세협상의 주요 의제가 되고 있어, 우리나라 입장에서는 美日 협상 과정 에서 對美 협상전략의 단서를 얻을 수 있을 것임 - (수출 전략의 재검토) 트럼프 2기 행정부는 무차별적 관세 부과, 우회 수출 제약 등 국내 기업이 지금까지 구축해 온 공급망으로부터의 이익이 훼손될 가능성도 큰 만큼 공급망 재편에서부터 수출 기지 선정에 이르기까지 전반 적인 수출 전략의 재검토도 필요함 ․ 수출 품목 및 지역의 다변화, 현지화 등을 통해 특정 공급망 또는 수출 기 지에 대한 의존도를 낮추는 한편 불확실성이 높은 글로벌 교역 환경에 대 해서도 대응할 수 있는 여건을 확보해야 함 ․ 이는 국내를 통한 타국 기업의 우회수출, 국내 기업의 타국 이용 우회수출 등도 트럼프 2기 행정부의 규제 대상이므로 발 빠른 대응이 필요함 ․ 인도, 아세안 등 성장잠재력이 높은 시장 공략을 수출의 외연 확대, 점진 적인 對中 공급망 재편 등을 통해 중장기 수출 기반 안정화를 꾀해야 함 셋째, 정책 이니셔티브 발현을 통해 민간투자를 기반으로 한 고용 및 소득 확충 을 유도하는 한편, 이것이 소비 확대로 이어질 수 있는 경제 선순환 구조 를 확립해야 한다. - (민간투자 기반 고용 및 소득 확충) 기업친화적 투자 환경 조성, 외국인 직접 투자 유치 확대, 국내 건설시장의 조기 안정화 등을 통해 민간투자를 기반으 로 한 고용 및 소득 확충 노력이 필요함 ․ 대외적으로는 트럼프 2기 행정부 관세정책 등에 의한 해외투자 가속, 대내 적으로는 고금리 등 고비용 리스크와 내수 여건 악화 등으로 민간투자 부 진 현상이 장기화되고 있음 ․ 이의 극복을 위해서는 적극적인 기업 친화적 투자 환경 조성 노력 등 국 내 투자 유인을 증대시키는 것이 긴요함 2025년 한국 경제 전망(수정) 29 ․ 구체적으로는 기업 관련 규제 완화 노력 지속, 금융·세제상 지원 강화, 노사 관계 안정, 기업인의 사기 진작 정책 등 다각적인 투자 활성화 정책 이 동반되어야 함 ․ 또한, 의료나 교육 등은 물론 수출입 규제 등 외국인투자 관련 인프라 전 반에 대한 개선, 유턴기업이나 유망 신기술 및 신산업에 관한 지원 강화 등의 노력이 필요함 - (건설시장 조기 정상화) 부동산 버블을 우려하지 않아도 될 정도 수준에서 국내 건설시장이 정상화될 수 있도록 유동성 공급 확대 및 합리적인 PF 시 장 관리, 실수요자에 대한 시장 접근성 강화 등의 시장 정책도 적극 추진되 어야 할 것임 넷째, 경기 침체기와 미래 불확실성 고조기에 가장 큰 타격을 받을 수 있는 취 약 계층에 대한 적절한 사회 안전망을 제공하여 사회 양극화가 고착화되 는 것을 막아야 한다. - (가계 특성별 사회 안전망 차별화) 저소득 고령 1인 가구 등 가계 특성별 사 회 안전망 차별화 등을 통해 복지 사각지대를 해소함과 동시에 복지재정의 효율성을 개선해야 함 ․ 현재 추진 중인 일자리나 금융 등의 부문에서의 사회 안전망의 경우에는 경기 회복세가 추세적으로 확인될 때까지 유지되는 것이 바람직함 - (업종별 영세 자영업자 지원 차별화) 영세 자영업자 비중이 높고 경기 침체 기 피해가 타 산업에 비해 상대적으로 큰 도소매, 음식숙박, 개인서비스 업 종의 경우, 업종별 규모별 피해 정도가 제각각인 만큼 차별적 지원 대책이 필요함 - (취약계층 대상 일자리 우선 제공) 청년층, 여성, 저소득층 등 고용 충격이 집중될 수 있는 계층에 대해서는 민간 일자리를 우선 지원할 수 있는 시스 템을 확충하고, 공공일자리도 관리·감독 강화를 통해 효율적이고 투명한 재 정 운영을 꾀해야 함 2025년 한국 경제 전망(수정) 30 다섯째, 사회통합을 기반으로 경제·사회의 역동성 복원 노력을 통해 중장기적 으로 한국 경제가 지속성장할 수 있는 기반을 확보해야 한다. - 등 실물경제를 중심으로 유연하게 운용하되, 그 과정에서 발생할 수 있는 가계부채 증가 등의 금융 불균형 문제는 금융정책으로 대응하는 정책조합을 고려할 필요 \n\n▣ 재정정책은 건전 재정기조를 훼손하지 않는 범위에서 취약계층 지원, 성장동력 강화를 위한 구조조정정책 등 필수적인 부분에 대한 정책을 확대할 필요 \n\n▣ 중장기적 관점에서 우리 경제가 전망\n\n ∙대내외 불확실성 확대로 민간소비 회복이 둔화되고 누적된 수주 부진의 영향으로 건설투자가 크게 부진한 가운데 미국 관세 인상 여파로 수출도 위축되면서 성장이 둔화될 전망\n\n▣ 민간소비 증가율은 0.9%를 기록할 전망\n\n ∙1/4분기 중 민간소비는 전기대비 0.1% 감소, 전년동기대비 0.5% 증가에 그쳐 부진이 심화\n\n ∙대내외적 불확실성의 해소가 지연됨으로 인해 소비심리는 장기간 위축될 전망\n\n ∙기업 경영 여건의 악화와 함께 고용시장이 둔화되는 등, 가계의 소득 여건 또한 연중 악화될 것으로 예상\n\n▣ 건설투자 증가율은 △5.7%를 기록할 전망\n\n ∙건설투자는 2022~2023년 수주 부진의 영향이 2025년까지 영향을 미치면서 △5.7%로 부진이 심화될 것으로 예상\n\n ∙2024년 2분기 이후 금리 인하 및 공사비 인상률 둔화 등으로 건설 수주가 완만한 회복세를 보여왔으나 그 수준은 여전히 낮은 상황\n\n ∙또한, 인허가 및 착공 물량이 여전히 감소세를 보이고 있는 점과 미분양 적체 등은 향후 건설투자 회복 속도를 제약하는 요인\n\n▣ 설비투자 증가율은 0.3%를 기록할 전망\n\n ∙상반기 예정된 설비투자는 일부 진행될 것으로 보이나, 관세 정책의 영향이 커지는 연말로 갈수록 설비투자 증가율 둔화폭이 커질 것으로 예상\n\n ∙다만, 관세 정책이 현재 예고된 수준보다 낮아질 가능성, 올해 상당수 항공기 도입이 예정되었던 점 등은 설비투자 증가율 하락을 제한하는 요인임.\n\n▣ 총수출 및 총수입 모두 0.3% 수준의 낮은 증가율을 기록할 전망\n\n ∙총수출은 미국의 관세 부과 및 경제 불확실성 고조에 따른 전 세계적 수요 위축으로 인해 증가율이 저조한 수준에 머물 전망\n\n* WTO는 2025년 전 세계 상품교역 성장률을 △0.2%로 전망하였는데, 이는 작년 10월에 발표된 전망치 2.9% 대비 3.1%p만큼 하향 조정된 수치에 해당\n\n ∙총수입은 글로벌 교역 위축이 중간재 및 자본재 수입에 대한 제약 요인으로 작용하는 가운데 소비심리 위축에 따라 내구성 소비재 수입도 부진한 모습을 나타내면서 낮은 증가율을 기록할 전망\n\n2. 고용\n\n▣ 2025년 고용률은 62.7%를 기록할 전망\n\n ∙고용률은 기업 경영 여건의 악화 및 내수 부진의 지속에도 불구하고 60세 이상 고령층의 견조한 노동시장 진입으로 전년도 수준을 유지할 전망\n\n ∙실업률은 전년대비 소폭 상승한 2.9%를 기록할 것으로 예상\n\n ∙취업자 수는 약 8만명 증가하여 전년대비 증가세가 둔화될 전망\n\n3. 물가 및 시장금리 \n\n▣ 소비자물가상승률은 상반기 2.0%, 하반기 1.9%로 연간으로는 2.0%를 기록할 전망\n\n ∙내수 부진 지속, 수출 감소 등으로 경제성장률이 크게 낮아지는 가운데 글로벌 경기침체 우려에 따른 유가 등 원자재가격 하락으로 물가는 하향 안정세를 유지할 것으로 전망\n\n ∙지정학적 위험 및 관세 협상 관련 불확실성 확대, 연준 금리인하 지연 등으로 인한 원/달러 환율 상승 가능성은 물가에 상방 압력으로 작용할 것으로 예상\n\n▣ 국고채 3년물의 연평균 금리는 2.3%로 전년 대비 상당폭 하락할 전망\n\n ∙작년말 이후 대내외 불확실성 확대로 경기 하방 리스크가 커지면서 기준금리에 대한 시장의 기대가 하향 조정되었고, 올해 상반기 중 시장금리도 지속적인 하락세 시현 \n\n ∙다만, 이미 국고채 3년물과 기준금리 간 역전폭이 큰 상황으로 이는 기준금리 인하 가능성이 상당 부분 선반영되어 있음을 시사\n\n ∙전년대비 증가 예정인 국고채 발행량과 추경 가능성, WGBI 편입 지연 등 수급 이슈 또한 금리 하락세를 일정 부분 제한할 것으로 예상\n\n4. 국제수지 전망\n\n▣ 경상수지는 880억 달러 흑자를 기록할 전망\n\n ∙전 세계 원유수요 둔화로 국제유가가 급락하면서 수출입물가도 큰 폭으로 하락함에 따라 수출입 모두 금액 기준으로 감소할 전망\n\n ∙2024년에 높은 수준을 기록했던 수출금액의 기저효과가 통관수출 감소율을 확대시키는 요인으로 작용하는 한편, 국제유가에 대한 탄력성 차이는 수입물가 및 수입금액이 수출 대비 상대적으로 더 큰 폭의 감소율을 기록하는 배경으로 작용\n\n ∙서비스 부문에서는 운송·관광 부문의 지급과 수입이 모두 동반부진을 경험하면서 서비스수지가 기존과 유사한 수준의 적자를 지속할 전망\n\nⅡ. 종합평가 및 정책제언\n\n▣ 2025년 세계 경제는 통상 마찰 격화와 정책 불확실성 확대로 성장과 교역이 둔화될 전망\n\n ∙IMF는 세계 경제성장률 전망치를 2025년 2.8%, 2026년 3.0%로 올해 1월 전망 대비 각각 0.5%p, 0.3%p 하향 조정\n\n ∙4.2일 발표된 미국의 보편관세 및 상호관세는 글로벌 금융시장에 충격을 주고 무역 관련 불확실성을 전례 없는 수준으로 높여 실물경기에 압박을 가하고 있으며, 상호관세 유예 발표 이후에도 이어지는 국가별 협상 관련한 불확실성이 지속\n\n* 미국 소비자 심리지표가 최근 급격히 둔화된 가운데 관세 회피를 위한 선제적 구매 효과가 사라진 이후의 소비·투자 둔화 위험이 확대\n\n ∙미국 정책 결정 과정의 즉흥성과 비일관성도 기업 투자와 민간 소비를 위축시키는 불확실성 요인으로 작용\n\n ∙관세로 인한 수입 물가 인상 압력이 현실화 할 경우 연준의 기준금리 인하 또한 지연될 가능성\n\n ∙이에 따라 IMF는 2025년 미국 경제성장률 전망을 1.8%로 1월 전망에 비해 0.9%p 하향 조정하였으며 상호관세의 주요 타겟이 된 중국의 성장률도 4.0%로 0.6%p 하향 조정\n\n* 세계 GDP의 43.4%를 차지(2024년 기준)하는 미ᐧ중 무역 갈등이 장기화 조짐을 보임에 따라 글로벌 경기둔화 위험이 현실화\n\n ∙중장기 성장률 전망 또한 약 3.2%로 과거 장기평균(2000-2019년, 3.7%) 대비 낮을 것으로 전망되어 구조적인 성장둔화 우려가 확대\n\n▣ 2025년 우리 경제는 민간소비, 건설투자를 중심으로 내수 부진이 지속되는 가운데 통상환경 악화로 순수출도 위축되면서 성장세가 당초 예상보다 크게 둔화될 전망\n\n ∙2024년 2분기 이후 GDP는 4분기 연속 전기비 0.1% 이하로 사실상 정체\n\n ∙건설투자는 누적된 수주 부진의 영향이 본격화하고 지방 부동산 시장 침체가 지속되면서 2024년 2분기 이후 4분기 연속 전기비 마이너스 성장을 기록\n\n* 건설투자를 제외한 2024년 및 2025년 1분기 성장률(전년동기비)은 각각 2.8%, 1.5%\n\n ∙당초 금리 인하 및 물가 안정으로 개선이 예상되었던 민간 소비는 대외불확실성 확대로 인한 경기둔화 우려 등에 따라 회복 속도가 둔화될 전망이며, 설비투자도 글로벌 경기 둔화로 성장세가 예상보다 낮아질 것으로 판단\n\n ∙수출은 미국 관세 인상과 수출시장 다변화를 위한 중국의 저가 공세 심화 등으로 증가세가 위축될 것으로 전망\n\n▣ 내수 부진과 수출 둔화가 맞물리며 성장세가 크게 약화된 최근 상황을 고려할 때, 보다 적극적인 경기 대응을 위한 방안을 강구하는 한편 중장기적 경제구조 개선을 위한 노력을 병행할 필요\n\n▣ 통화정책은 인플레이션 안정세와 경기 하방 압력 확대를 감안하여 보다 완화적으로 운용하고, 기준금리 조정 이외에 금융기관에 대한 신용공급 확대 등 대안을 모색\n\n ∙유가 안정세와 국내 수요 부진 심화로 추가적인 기준금리 인하 필요성이 높아진 상황이며, 내외금리차 역전 폭, 환율 변동성 등으로 기준금리 인하 폭이 제한되는 경우 경제지표와 연동한 포워드 가이던스 활용 등을 검토\n\n ∙경기 하강 지속 시 한은의 금융중개지원대출 등 금융기관에 직접 자금을 투입할 수 있는 방안을 모색함으로써 중소 수출기업 등에 대한 자금지원을 확대\n\n▣ 금융정책은 대외여건 변화와 금융시장 변동성 확대에 선제적으로 대응하는 한편 금리 인하 등으로 확대된 신용이 다시 부동산으로 쏠리지 않도록 경계하는 등 금융시장 안정 및 건전성 관리에 유의\n\n ∙미·중 무역갈등 격화와 글로벌 경기 둔화 우려 등으로 인한 급격한 자본유출 위험에 대비하여 금융시장 모니터링을 강화하고 비상대응체계를 유지\n\n ∙상환가능한 범위 내의 대출이라는 대원칙 하에 총부채원리금상환비율(DSR) 규제를 근간으로 한 가계부채 관리 기조를 일관되게 유지\n\n ∙다만, 경기 둔화로 기업 및 소상공인에 대한 신용 공급이 위축되지 않도록 관리\n\n▣ 재정정책은 중장기적 차원의 재정건전성 유지라는 현실적 제약을 감안하되 급격한 경기 둔화 시에는 보다 적극적이고 효율적으로 운용할 필요\n\n ∙경기 하방 압력 확대 시 신속하게 추가적인 추경예산 편성 및 집행을 고려하고, 경기 위축 시 피해가 예상되는 소상공인 및 저소득층 등 취약계층에 대한 선별적 지원을 통해 소비 위축을 최소화\n\n ∙관세 인상으로 피해가 예상되는 수출기업 및 산업에 대해 무역금융, 긴급유동성 지원 및 고용유지보조금 공급을 확대\n\n ∙이와 함께 미래 성장동력 확충을 위한 인프라 투자를 확대\n\n* 전기차 보급, 데이터센터, AI 모형 훈련 및 운용 등으로 크게 증가할 것으로 예상되는 전력수요에 대비하여 선제적 전력 인프라 투자를 실시\n\n** 국ᐧ공립 어린이집 및 초등학교 방과후 돌봄 확대, 유아 원격의료 및 상담 시스템 구축 등 육아 인프라 확대를 통해 저출산 심화에 대외 불확실성에도 불구하고 고금리 기조가 완화되는 가운데 반도체 관련 투자 수요도 양호한 흐름을 나타냄에 따라 금년(1.7%)과 내년(1.6%)에 완만한 증가세를 지속할 것으로 예상됨.\n*   **건설투자**는 작년(-3.0%)에 이어 금년(-4.2%)에도 감소세를 지속하겠으나, 건설수주 개선의 영향이 점차 반영되며 내년에는 2.4% 정도 증가하면서 부진이 완화될 전망\n*   **수출**은 미국 관세인상에 따른 세계교역 위축으로 둔화될 것으로 전망\n*   **경상수지**는 교역조건 개선에도 불구하고 수출 부진으로 흑자폭이 축소될 전망\n*   **취업자 수** 증가폭은 작년 16만명에서 금년 9만명, 내년 7만명 수준으로 축소될 전망\n\n**Ⅰ. 현 경제상황에 대한 인식**\n\n*   **□ 최근 우리 경제는 대내외 불확실성이 확대되면서 성장세가 둔화되는 모습**\n\n*    · 1/4분기 국내총생산은 전년동기대비 0.1% 감소하였으며, 계절조정 전기대비로도 4분기 연속 0% 내외의 증가율을 기록하며 정체되는 모습\n*    · 경제활동별로는 건설업이 대폭 감소한 가운데 제조업과 서비스업의 증가세도 둔화\n\n*   **□ 내수는 정국 불안에 따른 심리 위축이 지속되고 대외 불확실성이 확대되면서 가시적인 회복세가 나타나지 않는 모습**\n\n*    · 소비자심리 위축으로 숙박⋅음식 등 서비스를 중심으로 민간소비 증가세가 둔화\n*    · 설비투자가 양호한 증가세를 보였으나 대외 불확실성 확대로 기업심리가 위축되는 등 투자 여건이 악화되고 있으며, 건설투자는 감소세가 확대되며 부진이 심화\n*    ·경기 부진에 따라 물가상승세가 2% 내외를 지속한 가운데, 고용증가세는 건설업과 제조업에서 감소세가 심화되면서 완만한 수준에 머무름.\n*    ·한편, 그동안 내수경기에 하방 압력으로 작용해 왔던 고금리 기조와 건설수주 부진이 일부 해소되고 있어 향후 시차를 두고 내수 부진이 다소 완화될 가능성을 시사\n\n*   **□ 수출은 최근까지 반도체의 호조세에도 불구하고 여타 산업의 부진으로 둔화되고 있으며,향후에는 미국 관세인상으로 수출 여건이 더욱 악화될 가능성**\n\n*    · 상품수출(물량)이 감소한 가운데 경기 부진으로 상품수입(물량) 증가세도 둔화\n*    · 대규모 경상수지 흑자 추세가 지속되고 순대외자산 규모도 GDP 대비 60% 수준까지 증가하는 등 대외건전성은 양호한 모습\n*    · 4월 들어 미국이 관세를 대폭 인상하였으며, 통상 관련 불확실성도 급격히 상승\n*    · 최근 IMF는 통상분쟁의 부정적 영향을 반영하여 세계경제 성장률을 미국과 중국을 중심으로 대폭 하향 조정하였으며, 글로벌 경기 둔화로 국제유가도 하락세를 보임.\n\n*   **□ 이와 같은 대내외 여건을 고려할 때, 우리 경제는 통상 여건 악화에 따른 수출 부진으로 성장세가 둔화될 전망**\n\n*    · 내수가 여전히 미약한 증가세에 머물러 있으나, 금리 하락세, 소비심리 개선 및 건설 수주 증가 등이 반영되면서 향후 부진이 다소 완화될 것으로 예상됨.\n*    ·그러나 미국이 광범위한 품목에 대해 관세를 인상한 가운데, 관련 불확실성도 이례적인 수준으로 확대되며 수출 여건이 급격히 악화\n\n*   **□ 이러한 경기 둔화 흐름을 감안하여 거시정책은 완화적 기조로 운용하는 것이 바람직**\n\n*    · 대내외 수요 둔화로 초래될 수 있는 물가 하방 압력을 축소하기 위해 통화정책은 보다 완화적인 기조로 운용하는 것이 바람직\n*    · 재정정책은 큰 폭의 관리재정수지 적자(86.4조원, GDP 대비 3.3%)를 감안하면 이미 어느 정도 완화적 기조로 편성되어 있는바, 정부지출 추가 확대에는 신중하게 접근\n\n*   **□ 이와 함께 잠재성장률 하락세를 감안하여, 경제 구조개혁을 위한 정책적 노력을 지속**\n\n*    · 금년 우리 경제의 잠재성장률은 1%대 후반으로 추정되며, 2040년대에는 잠재성장률이 0% 내외까지 하락할 전망\n*    · 따라서 진입장벽과 노동시장 경직성을 완화하는 등 생산성 개선을 위한 노력이 필요\n*    · 이와 함께 거시정책 기조도 이러한 잠재성장률 하락 추세를 감안하여 설정할 필요\n\n**Ⅱ. 2025~26년 국내경제 전망**\n\n**1. 대외여건에 대한 주요 전제**\n\n*   **□ 2025~26년 세계경제는 국제 무역이 위축됨에 따라 성장세가 약화될 것으로 전제**\n\n*    · 최근 IMF는 미국의 보호무역주의가 초래한 통상환경 악화로 2025년 세계경제가 2024년(3.3%)보다 낮은 2.8%의 성장률을 기록한 후, 2026년에도 3.0% 성장에 그칠 것으로 전망\n*    · 미국 관세정책은 기본관세 10%와 현재의 품목별 관세율이 향후에도 지속될 것으로 전제\n\n*   **□ 2025년 원유 도입단가(두바이유 기준)는 수요 부진에 기인하여 2024년(배럴당 80 달러)보다 낮은 69달러를 기록한 후 2026년에는 66달러로 소폭 하락할 것으로 전제**\n\n*   **□ 실질실효환율로 평가한 원화가치는 최근 수준에서 큰 변동이 없을 것으로 전제**\n\n**2. 2025~26년 국내경제 전망**\n\n*   **□ 우리 경제는 2025년에 건설업 부진과 통상 여건 악화로 0.8% 성장하는 데 그친 후,2026년에는 통상분쟁의 여파에도 불구하고 완만한 내수 회복으로 1.6% 성장할 전망**\n\n*    · 민간소비는 금년에도 1.1% 내외의 낮은 증가율을 기록하겠으나, 정국 불안의 영향이 완화되고 금리 인하도 반영되면서 내년에는 1.6%의 완만한 회복세를 나타낼 전망\n*    · 설비투자는 높은 대외 불확실성에도 불구하고 고금리 기조가 완화되는 가운데 반도체 관련 투자 수요도 양호한 흐름을 나타냄에 따라 금년(1.7%)과 내년(1.6%)에 완만한 증가세를 지속할 것으로 예상됨.\n*    · 건설투자는 작년(-3.0%)에 이어 금년(-4.2%)에도 감소세를 지속하겠으나, 건설수주 개선의 영향이 점차 반영되며 내년에는 2.4% 정도 증가하면서 부진이 완화될 전망\n*    · 수출은 미국 관세인상에 따른 세계교역 위축으로 둔화될 것으로 전망\n*    · 경상수지는 교역조건 개선에도 불구하고 수출 부진으로 흑자폭이 축소될 전망\n\n*   **□ 소비자물가는 경기 둔화 및 유가 하락으로 낮은 상승률을 지속할 것으로 전망**\n\n*    · 소비자물가는 금년에 1.7% 상승한 후, 내년에는 국제유가 하락폭이 축소되고 내수도 완만한 회복세를 나타내면서 1.8% 정도 상승할 전망\n*    · 근원물가(식료품 및 에너지 제외)도 금년에 1.8% 수준의 낮은 상승률을 기록한 후,내년에는 1.9%로 상승폭이 소폭 확대될 것으로 예상됨.\n\n*   **□ 인구구조 변화가 지속적인 하방 압력으로 작용하는 가운데, 불확실성 증대에 따른 고용 여건 악화로 취업자 수 증가폭은 작년 16만명에서 금년 9만명, 내년 7만명 수준으로 축소될 전망**\n*   \n · 실업률은 작년 2.8%에서 금년과 내년에 각각 3.0% 수준으로 소폭 상승할 전망\n\n**3. 전망의 위험요인**\n\n*   **□ 우리 경제는 미국 관세정책과 여타 국가들의 대응에 따라 평소에 비해 상당히 높은 불확실성에 직면하고 있음.**\n\n*    ·미국이 높은 관세율을 유지하는 가운데 상대국들이 보복관세로 대응하며 통상분쟁이 격화되는 경우, 우리 경제의 성장에도 추가적인 하방 압력으로 작용할 것으로 예상됨.\n*    ·반면, 협상이 원만히 타결되면서 수출 여건이 빠르게 개선될 가능성도 존재\n\n*   **□ 대내적으로는 주택경기가 하락하면서 건설업체들의 재무건전성이 악화되는 경우, 건설 투자 회복이 지연될 가능성**\n\n*    · 비수도권에서 준공 후 미분양이 급증하는 가운데 주택매매가격이 하락세를 지속하는 등 주택경기가 하락하고 있음.\n*    · 이러한 상황에서 부동산 PF 대출 구조조정이 지연되고 건설업체 재무건전성이 추가적으로 악화될 경우, 공사진행에 차질이 발생하면서 건설투자 회복이 제약될 수 있음.\n\n![Image 13](https://www.kdi.re.kr/upload/publication/18775/editor/A07E10A01B_20250514175922.png)\n\n**Ⅲ. 정책방향**\n\n**1. 재정정책**\n\n*   **□ 최근 큰 폭의 관리재정수지 적자가 지속되고 있음을 감안할 때, 재정건전성 유지를 위해 추가적인 재정지출에는 신중하게 접근하는 것이 바람직**\n\n*    · 코로나19 위기가 지나간 2023~24년에도 큰 폭의 관리재정수지 적자가 발생\n*    · 추가경정예산이 확정됨에 따라 2025년에도 관리재정수지 적자가 GDP 대비 3.3%로 상향 조정되었으며, 경기 둔화에 따른 세입 여건 악화를 감안하면 관리재정수지 적자 폭이 더욱 확대될 가능성도 존재\n*    · 이와 같이 예산이 이미 확장적 기조로 편성되어 있다는 점을 감안하여 정부지출 추가 확대는 신중을 기하는 것이 바람직\n\n*   **□ 잠재성장률 하락에 따른 세입 여건 악화와 국민연금 지급보장 법제화 등을 감안하여,재정건전성이 크게 훼손되지 않도록 사전적으로 노력할 필요**\n\n*    · 국민부담률이 빠르게 상승하고 있음에도 불구하고 인구구조 변화에 따른 중장기적 성장세 둔화로 향후 국가채무가 급증할 것으로 예상되고 있음.\n*    · 국민연금법 개정에도 불구하고 향후 기금 고갈이 예상되며, 국가의 연금급여 지급보장이 명문화됨에 따라 미래세대의 부담이 크게 확대될 가능성\n*    · 이와 같이 중장기적으로 재정 여건이 크게 악화될 수 있음을 감안하여, 재정적자가 지나치게 확대되지 않도록 사전적으로 제도를 개편할 필요\n\n**2. 통화정책**\n\n*   **□ 통화정책은 물가 하방 압력에 대응하여 보다 완화적인 기조로 운용하는 것이 바람직**\n\n*    · 최근 경기 둔화에 따른 물가 하방 압력이 존재하며, 대외 불확실성 확대로 향후 물가 하방 압력이 확대될 가능성\n*    · 대외건전성이 양호한 가운데 금융시스템 위험도 높지 않을 것으로 판단되므로, 통화정책은 물가 안정에 중점을 두고 운용할 수 있을 것으로 사료됨.\n*    · 이러한 점을 종합적으로 고려할 때, 통화정책은 수요 둔화로 초래될 수 있는 물가 하방 압력을 축소할 수 있도록 완화적으로 운용하는 것이 바람직\n\n**3. 금융정책**\n\n*   **□ 대내외 불확실성이 높은 상황임에도 금융시스템은 대체로 안정적인 모습**\n\n*    · 주요 금융업권의 건전성 지표가 규제 수준을 상회하는 등 단기적으로 금융시장의 시스템리스크 가능성은 높지 않은 것으로 판단됨.\n*    · 2단계 스트레스 DSR 규제 시행, 가계대출 금융감독 등 거시건전성 정책 강화에 따라 GDP 대비 가계부채 비율은 완만한 하락 추세를 지속\n*    · 이와 함께 국가 부도위험을 나타내는 CDS 프리미엄이 낮은 수준을 유지하고 있으며,연간 1,000억달러 내외의 대규모 경상수지 흑자와 GDP 대비 60% 수준의 순대외자산을 감안하면 대외건전성도 비교적 양호한 모습\n\n*   **□ 구조적으로 생존 가능성이 희박한 부실 기업에 대한 금융지원을 자제하고 거시건전성 정책을 일관되게 추진하는 것이 바람직**\n\n*    · 기준금리 인하 기조, 부동산 PF 구조조정 인구 비중이 빠르게 증가함에 따라 평균소비성향이 점차 반등할 것으로 예상됨. - 지난 20년간 기대수명은 약 6.5세 증가한 반면, 향후 20년간의 기대수명 증가는 절반 수준인 약 3.5세에 그칠 것으로 전망 - 저출생 기조가 지속되며 고령층 인구의 비중이 빠르게 증가하면 생산가능인구가 축소되며 성장률이 더욱 하락하는 가운데 소비성향은 점차 증가할 것으로 예상 - 따라서 민간소비 증가세가 경제성장률을 하회하는 현상도 점차 완화될 전망 ■ 지난 20년간의 평균소비성향 하락에는 우리나라 노동시장의 구조적인 문제가 투영되어 있는 만큼, 이를 완화하기 위한 대응 방안을 모색할 필요 - 기대수명 증가에 대응하여 은퇴 시점이 적절히 조정될 수 있도록 고령층의 노동시장 참여를 제약하는 구조적 요인들을 해소할 필요 - 따라서 연공서열형의 경직적인 임금구조를 개선하고 직무 및 성과 중심의 임금체계를 강화하는 한편 정년퇴직 후 재고용 제도를 활성화하는 등 노동시장의 마찰적 요인을 해소하여 고령층 노동 수요를 확대할 필요 - 특히 고령 인력의 적절한 활용이 확대될 경우 저출생·고령화에 따른 잠재성장률 하락 압력을 부분적으로 상쇄할 수 있을 것으로 판단됨.](https://www.kdi.re.kr/research/analysisView?art_no=3669)\n\n    *   ![Image 27](https://www.kdi.re.kr/data/download/resumes/img/picture/20180702.jpg)_프로필_ **김미루 국채연구팀장**프로필 보기\n    *   ![Image 스크립트  | 열기\n\n 마스크를 벗은 지도 벌써 3년째지만,\n\n 한국경제는 여전히 코로나19 위기 이전으로 회복하지 못하고 있습니다.\n\n 올해도 성장률이 1% 수준에 그칠 거라는 걱정이 많죠.\n\n 우리나라의 잠재성장률, 그러니까 성장할 수 있는 능력치 자체가 계속 떨어져서\n\n 본격적으로 저성장 시대에 들어선 건 아닐까요?\n\n KDI가 한국의 잠재성장률을 전망해봤습니다.\n\n 우선 한 나라의 경제성장은 노동과 자본 같은 생산요소를 얼마나 투입하는지,\n\n 그리고 이 두 가지를 얼마나 효율적으로 활용하는지를 나타내는\n\n 총요소생산성으로 결정됩니다.\n\n 기술 개발이나 조직 혁신 같은 노력으로 생산성을 더 높일 수 있죠.\n\n 지난 20년의 경제성장을 분석해보니\n\n 2000년대에 비해 2010년대 경제성장률이 1.6%p 하락했는데,\n\n 이 중 약 67%는 생산성 증가세가 둔화되었기 때문입니다.\n\n 그리고 최근 10년 동안 경제성장률이 더 떨어진 건\n\n 생산성뿐만 아니라 노동 투입, 즉 일할 사람이 예전만큼 늘어나지 못했기 때문이었습니다.\n\n 실제로 생산연령인구는 2019년에 정점을 찍은 뒤 매년 수십만 명씩 줄고 있습니다.\n\n 앞으로 점점 줄어 2050년에는\n\n 생산연령인구 비중이 전체인구의 절반밖에 안 될 거라고 합니다.\n\n 반면, 65세 이상 고령인구의 비중은 올해 20%를 넘어서고,\n\n 2050년에는 40%까지 오를 전망이고요.\n\n (인터뷰 – 김지연 박사님)\n\n 문제는 이러한 인구구조의 변화가 노동투입뿐만 아니라\n\n 생산성에까지 악영향을 줄 수 있다는 점입니다.\n\n 청년층은 고령층에 비해 새로운 기술에 대한 적응력이 높습니다.\n\n 또한 청년층의 참신함과 위험을 감수하려는 성향은 혁신의 원동력이기도 한데요.\n\n 따라서 청년들의 비중이 줄어들면 우리나라 전체 생산성에 부정적인 영향을 줄 수 있습니다.\n\n 앞으로 우리 경제 전반의 생산성이 얼마나 좋아질 수 타격을 주고 있습니다. 현대차의 미국 수출 물량은 최근 20% 감소하였으며, 이는 2020년 코로나19 사태로 인한 선적 감소 이후 처음으로 나타난 현상입니다. 이러한 관세 부과로 인해 현대차의 연간 영업이익이 약 5조 원 감소할 것이라는 분석까지 등장하고 있습니다. 저렴한 가격대의 차종이 아닌 고가 차종 위주로 미국 시장에 진출한 현대차로서는 가격 경쟁력에 심각한 도전 과제를 안고 있습니다.\n\n*   현재 현대차는 조지아주 신공장의 생산 능력을 증가시켜 미국 내 생산 비중을 높이고 관세 비용 상승세에 대응하고자 노력하고 있습니다. 하지만 관세가 본격적으로 적용되는 오는 6월과 7월 사이의 판매 실적은 예상보다 부진할 가능성이 높습니다. 미국 소비자들이 이미 관세 부과 전 차량 구매를 몰아간 점에서, 2분기 매출은 플러스 성장이 사실상 어려울 것으로 분석되고 있습니다.\n\n*   ### 3-2. 원·달러 환율 변동 추이\n\n*   원·달러 환율은 현대차 실적에 중요한 변수로 작용하고 있습니다. 2025년 1분기 평균 환율이 1453원이었던 반면, 최근 환율은 1380원대로 급락하면서 현대차의 실적에 부정적인 영향을 미치고 있습니다. 해외 판매 비중이 높은 현대차는 원화 강세가 이어질 경우 평균 판매 단가(ASP)의 감소를 피할 수 없습니다. 이로 인해 향후 실적에 대한 기대치는 점점 저조해지고 있습니다.\n\n*   예를 들어 한국투자증권은 현대차의 ASP가 달러 기준으로 이미 역성장에 가까운 수치를 기록할 것으로 보며, 이번 원화 강세가 지속되면 추가적인 실적 부진이 우려된다고 밝혔습니다. 지난 1분기 환율 상승이 긍정적인 영향을 미쳤지만, 이제는 급변하는 외부 환경에 의해 역풍을 맞을 날이 멀지 않았다는 실상을 반영하고 있습니다.\n\n*   ### 3-3. 주요 원자재 가격·노조 리스크\n\n*   원자재 가격 변동 역시 현대차의 경영에 영향을 미치고 있습니다. 금속 가격과 반도체 수급 부진이 겹쳐 부품 생산비용 상승을 초래하고 있으며, 이는 곧 차량의 총 제조 단가에 반영되고 있습니다. 이러한 상황에서 현대차는 추가적인 비용 절감을 위해 공급망 최적화와 현지 물량 조정을 강구하고 있으나, 현실적으로 해결해야 할 과제가 많습니다.\n\n*   또한, 올 여름 시작될 임단협은 또 다른 불안 요소로 자리 잡고 있습니다. 노조는 이번 협상에서 대규모의 임금 인상과 근무 조건 개선을 요구할 것으로 예상되고 있으며, 회사의 재정적 여건이 여의치 않은 만큼 갈등은 고조될 가능성이 큽니다. 현대차는 노조와의 협상에서 관세와 환율 등 외부 리스크가 가시화된 상황에서 내부 체질 개선의 필요성을 강조하며 균형을 찾아가려는 노력을 기울이고 있습니다.\n\n4. 1분기 실적 리뷰 및 주요 동인\n--------------------\n\n*   현대차의 2025년 1분기 실적은 역대급 성과로 기록되었습니다. 매출 37조7787억원, 영업이익 3조5927억원, 순이익 3조4194억원이라는 수치는 모두 전년 동기 대비 큰 폭으로 성장한 것입니다. 특히, 이러한 성장은 SUV 및 친환경차의 판매 증대와 함께 평균판매단가(ASP)의 향상에 기인하고 있습니다. 이는 단순한 숫자를 넘어 현대차의 전략적 유연성과 시장 내 위치를 잘 보여주는 사례입니다.\n\n*   ### 4-1. 1분기 매출·영업이익 실적 요약\n\n*   2025년 1분기 현대차는 총 102만1712대를 판매하며, 이는 전년 동기 대비 13.2% 증가한 수치입니다. 특히 내수 시장에서의 판매는 25.6% 상승하여 19만1047대를 기록했습니다. 이러한 판매 성장은 도심형 SUV 및 제네시스 브랜드의 누적 판매 증가로 이어지고 있으며, 이에 따라 평균판매단가가 3조1360억원의 매출 증대 효과를 초래하였습니다. 매출 원가율 또한 개선되어 79.6%로, 이는 부품 공급 상황의 개선과 환율 효과 덕분입니다.\n\n*   이는 전반적인 판매 성장을 바탕으로 하였으며, 현대차의 전략적 계획이 성공적으로 실행되고 있음을 상징합니다. 그러나 이러한 긍정적인 성과 뒤에는 지정학적 리스크와 금리 인상으로 인한 수요 감소 우려가 함께하고 있음을 주의할 필요가 있습니다.\n\n*   ### 4-2. SUV·친환경차 판매 믹스 개선\n\n*   현대차의 1분기 판매 믹스에서 SUV와 친환경차의 비중 상승은 주목할 만한 변화입니다. SUV 판매 비중은 지난해 57.2%에서 57.8%로 증가하였고, 친환경차의 판매는 전년 대비 38.8% 상승하여 총 16만1000대에 달했습니다. 이 같은 변화는 고객의 환경 친화적 요구에 부응하는 동시에 기업의 지속 가능성 전략에도 부합합니다.\n\n*   특히, 고부가가치 차종의 판매 증대는 ASP 상승으로 이어졌고, 이는 현대차의 수익성을 더욱 강화하는 요소로 작용하고 있습니다. 고객의 선호도 변화에 발맞춘 모델 개발과 마케팅 전략 덕분에 현대차는 시장에서의 경쟁력을 계속해서 확장할 수 있게 되었습니다.\n\n*   ### 4-3. 환율 상승 효과 및 분기 배당 발표\n\n*   이번 분기에서는 원달러 평균 환율이 1276원으로 전년 대비 5.9% 상승하면서 매출 증대에 기여하였습니다. 환율 상승은 해외 매출을 수익성 있게 전환시키는 데 중요한 역할을 했으며, 이러한 요소는 현대차의 글로벌 시장 내에서 경쟁력을 강화하는 데 긍정적인 영향을 미쳤습니다.\n\n*   또한, 현대차는 주주 가치를 최대화하기 위해 신규 배당 정책을 발표했습니다. 기존 잉여현금흐름(FCF)을 기준으로 하던 배당 기준을 연결 지배주주 순이익으로 변경하였고, 이제는 분기별로 배당을 실시하기로 결정한 것입니다. 이러한 변화는 투자자들에게 보다 예측 가능하고 투명한 배당을 제공하여, 장기 보유의 매력을 더욱 강화할 것으로 기대됩니다.\n\n5. 2분기 실적 전망 시나리오\n-----------------\n\n*   2025년 2분기 현대차의 실적 전망은 관세와 환율, 그리고 내부적인 판매 모멘텀 간의 상호작용에 의해 크게 좌우될 것으로 예상됩니다. 특히 미국에서의 관세 부과의 여파가 크기 때문에, 이를 준비하고 대응하는 전략이 무엇보다 중요합니다. 이러한 변수들은 현대차의 매출 및 영업이익에 중대한 영향을 미칠 수 있으며, 다양한 시나리오를 통해 이를 분석해 볼 필요가 있습니다. 이제 각각의 시나리오를 살펴보겠습니다.\n\n*   ### 5-1. 낙관 시나리오: 현지 생산 확대·재고 활용으로 비용 축소\n\n*   낙관적인 시나리오에서는 현대차가 현지 생산을 확대하고 재고를 효과적으로 활용함으로써 비용 절감을 이룰 것으로 보여집니다. 현대차는 현재 미국 내 생산 라인을 더욱 강화하고 있으며, 이를 통해 관세의 영향을 최소화할 전략을 모색 중입니다. 이러한 전략은 이미 GM과 토요타가 경험한 바와 같이, 생산 물량을 조정함으로써 부정적인 영향을 완화할 수 있는 가능성이 큽니다. 더불어, 하이브리드차와 같은 친환경 모델의 판매 증가는 안정적인 수익 기반을 제공할 것으로 보이며, 중장기적으로는 이익률 개선에 기여할 수 있습니다. 실질적으로 미국 내 고부가가치 차종에 대한 수요가 증가하며, 높은 가격대의 모델에서 판매 증가가 나타날 경우, 현대차는 관세 부담을 효과적으로 상쇄할 수 있을 것입니다. 이러한 긍정적인 요소들이 복합적으로 작용한다면, 현대차는 까다로운 매크로 환경 속에서도 희망적인 실적을 달성할 가능성이 높아지게 됩니다.\n\n6. 투자 전략 및 결론\n-------------\n\n*   현대차의 진정한 도전은 2025년의 글로벌 환경에서 어떻게 흔들림 없이 성장 가능성을 유지하느냐에 달려 있습니다. 관세와 환율이라는 외생 변수들이 영업 실적에 미치는 영향이 더욱 심각해지는 시점에서, 신중한 투자 전략은 그 어느 때보다도 중요합니다. 이러한 맥락에서, 현대차의 2분기 실적 전망 및 이를 반영한 투자 전략이 주목받고 있습니다.\n\n*   ### 6-1. 목표주가·투자의견 요약\n\n*   현재 현대차의 목표주가는 27만원으로 하향 조정되었습니다. 이는 미국의 25% 관세 부과가 미치는 영향을 반영한 것으로, 2025년의 영업이익은 기존 예상 대비 12% 감소한 12조5000억원이 될 것으로 추정됩니다. 이 보고서에서는 현대차의 PER(주가수익비율)가 4배에 불과하며, 이는 주가 하락에 대한 어느 정도의 방어력을 제공할 것입니다. SK증권에서는 투자 의견을 \'매수\'로 유지하였으며, 가격 조정의 부담에도 불구하고 향후 가파른 회복세를 기대하고 있습니다.\n\n*   ### 6-2. 배당수익률과 자사주 매입 효과\n\n*   현대차는 주주 가치를 증대시키기 위한 다양한 전략을 채택하고 있습니다. 자사주 매입과 배당은 단기적으로 주가를 지탱하는 역할을 할 것입니다. 예상 배당수익률은 약 3%로 보이며, 자사주 매입은 주가 상승 요인으로 작용할 수 있습니다. 과거에도 현대차는 안정적인 배당 정책을 통해 투자자들에게 매력적인 투자처로 자리 잡아 왔습니다. 현재 이러한 정책이 강화되면 안정적인 수익을 추구하는 투자자들에게 긍정적인 신호로 작용할 것입니다.\n\n*   ### 6-3. 관세 협상 전개 가능성 및 주가 모멘텀\n\n*   현대차의 성장모멘텀은 관세 협상 진행 상황에도 크게 의존하고 있습니다. 미국과의 협상에서 완화된 조건이 성립된다면, 영업이익의 하향 압력을 덜 수 있는 가능성이 있습니다. 최근의 재료비 상승과 규제 변화가 주가의 추가적인 하락을 유발할 수 있으나, 긍정적인 협상 결과는 주가 회복을 이끌 수 있습니다. 따라서, 관세 이슈 및 국제 정세의 변화를 면밀히 주시하는 것이 중요합니다.\n\n결론\n--\n\n*   현대차의 2025년 2분기 실적 전망은 관세와 환율 변동이라는 외부 요인의 압박 속에서 상당히 불확실한 상태에 놓여 있습니다. 1분기 성과는 긍정적이었지만, 2분기 실적에 우려가 커지는 가운데, 현대차는 이러한 난관을 극복하기 위한 다양한 전략을 강구해야 할 시점에 있습니다. 특히, 현지 생산 확대와 하이브리드의 결합 전략\n\n*   현대차·기아는 전기차(BEV)와 하이브리드(HEV) 차량의 생산을 혼류생산 방식으로 전환하여 양쪽 시장에서의 경쟁력을 강화하고 있습니다. 이를 통해 기존 공장에서 전기차와 하이브리드를 동시에 생산함으로써, 제조 효율성을 높이고 시장 변화에 즉각적으로 대응할 수 있는 체제를 구축하고 있습니다. 2023년에는 미국 시장에서 전년 대비 52.3% 증가한 27만8122대의 친환경차를 판매하였으며, 이러한 추세는 앞으로도 지속될 것으로 보입니다.\n\n*   또한, 현대차와 기아는 미리 계획된 신차 출시에 힘입어 고객의 선택의 폭을 넓히려 하고 있습니다. 현대차의 아이오닉7과 기아의 EV9과 같은 대형 전기차는 물론, 더 작은 차급의 전기차도 추가하여 다양한 소비자 요구를 충족할 계획입니다. 이와 같은 전략은 전기차 수요가 증가하는 가운데 다양한 차급의 차량을 제공함으로써 현대차·기아의 시장 점유율 확대에 긍정적인 영향을 미칠 것입니다.\n\n*   ### 2-3. 시장 내 경쟁력 강화 방안\n\n*   현대차·기아는 미국 시장 내 경쟁력 강화를 위해 다양한 전략을 채택하고 있습니다. 우선, 인플레이션 감축법(IRA) 시행에 따른 보조금 혜택이 축소된 상황에서도 경쟁력을 유지하기 위해 현지에서의 생산을 강조하고 있습니다. IRA 규정에 따라 현지에서 생산된 전기차는 세액 공제를 받을 수 있는데, 이를 통해 가격 경쟁력을 확보할 수 있습니다.\n\n*   또한, 현대차·기아는 고객에게 더 나은 인센티브를 제공하여 판매를 촉진하고 있는 점도 특징적입니다. 특히 미국 내 판매 목표를 달성하기 위해 판매 장려금(인센티브) 정책을 강화하고 있으며, 이는 소비자들에게 더 유리한 조건으로 차량 구매를 유도하는 역할을 하고 있습니다. 이렇게 만들어진 판매 환경은 결국 수익성 강화로 이어질 것으로 기대됩니다.\n\n*   마지막으로, 현대차·기아는 지속적인 연구 개발과 시장 조사로 변화하는 소비자 트렌드에 맞춰 제품과 서비스를 개선해 나가고 있습니다. 이는 브랜드 이미지 향상은 물론, 소비자 로열티를 증가시키는 데 중요한 요소로 작용할 것입니다.\n\n3. 전기차 생산의 본격화와 그 의미\n--------------------\n\n*   ### 3-1. 미국 내 전기차 생산기지 현황\n\n*   현대차·기아는 미국 시장에서 전기차 생산을 본격화하고 있으며, 이는 국내외 시장에서의 경쟁력을 강화하기 위한 중요한 전략입니다. 이러한 전략의 일환으로 현대차·기아는 조지아주와 앨라배마주에 전기차 생산 공장을 운영하고 있습니다. 특히, 조지아주 웨스트포인트 공장에서는 기아의 EV9과 같은 대형 전기 SUV 모델을 생산할 예정인데, 이는 미국 내 수요에 즉각 대응할 수 있는 체계를 구축하는 것을 목표로 하고 있습니다.\n\n*   또한, 현대차 그룹에서 설계한 'HMGMA(현대자동차그룹 메타플랜트 아메리카)'는 연간 30만 대의 전기차를 양산할 수 있는 규모를 목표로 하고 있어, 이는 북미에서의 전기차 생산 확대를 실현하는 기반이 될 것입니다. 현대차는 아이오닉5와 같은 모델을 이 공장에서 첫 생산할 계획이며, 이는 소비자들에게 더 많은 품질 높은 전기차 모델을 제공할 수 있는 기회를 열어줍니다.\n\n*   ### 3-2. 인플레이션 감축법(IRA)의 영향\n\n*   인플레이션 감축법(IRA)은 미국 연방 정부가 환경 친화적인 자동차의 구매를 장려하기 위해 시행한 법률로, 이를 통해 소비자들에게 세액 공제를 제공합니다. 현대차·기아는 이러한 법을 활용하여 미국 내 전기차 판매를 확대하려 하고 있습니다. 그러나 IRA 규정이 더욱 강화됨에 따라 보조금 혜택을 받기 위한 자격을 갖춘 차량의 수가 감소한 상황에서, 현대차·기아는 이를 기회로 삼아 현지 생산을 늘리고 있습니다.\n\n*   2024년 1월부터 시행된 IRA의 세부 조항에 따라 미국 내에서 판매되는 전기차는 반드시 미국 내 생산 공장에서 제조되어야 하며, 이는 현대차·기아가 미국 내에서의 공장 가동을 강화하고 현지 생산 비율을 2030년까지 75%까지 높이겠다는 전략적 목표를 수립하는 데 기여하고 있습니다. 이처럼, 현대차·기아는 실시간으로 변동하는 시장 환경에 발 빠르게 대응하며, 전기차 시장에서의 우위를 지속적으로 확보하고 있습니다.\n\n*   ### 3-3. 전기차 모델 출시 계획\n\n*   현대차·기아는 미국 시장에 맞춘 다양한 전기차 모델을 출시할 계획입니다. 기아는 EV9과 더불어 EV3, EV4 모델을 연내에 출시할 예정이며, 이러한 모델들은 소비자들에게 다양한 차급과 가격대의 선택지를 제공합니다. 이와 함께, 현대차는 아이오닉7을 출시하여 세 번째 전기차 전용 플랫폼을 적용한 모델로 시장에 도입할 계획입니다.\n\n*   이러한 다양한 모델 출시는 시장 내 경쟁력을 높이는 데 중요한 요소로 작용할 것입니다. 특히, 소비자들이 선호하는 느낌의 SUV와 크로스오버 모델 중심으로 라인업을 강화함으로써, 현대차·기아의 전기차 제품군은 더욱 다양해질 것입니다. 이러한 전략적 접근은 변화하는 트렌드에 맞춰 전기차에 대한 소비자의 수요를 극대화하고, 향후 성장 가능성을 더욱 높일 수 있을 것입니다.\n\n4. 하이브리드 시장의 급성장과 현대차·기아의 대응\n----------------------------\n\n*   ### 4-1. 미국 하이브리드 시장 성장 추세\n\n*   미국 하이브리드 자동차 시장은 최근 몇 년간 눈에 띄는 성장세를 보여줍니다. 2022년에는 전체 자동차 판매량 중 하이브리드 차량의 비중이 5%에서 8%로 상승하였고, 2023년에는 하이브리드의 판매 증가율이 71.3%에 달했습니다. 이는 지난해 1분기 전기차 판매 증가율이 62.6%에서 4분기에는 27.2%로 둔화된 것과 대조적입니다. 하이브리드는 이제 소비자들에게 친환경성과 연비를 제공하는 매력적인 대안으로 각광받고 있으며, 현대차·기아는 이러한 트렌드를 활용하여 시장 점유율을 확대하고 있습니다.\n\n*   ### 4-2. 경쟁 업체 분석 및 대비 전략\n\n*   현대차·기아는 미국 시장의 하이브리드 판매 증가에 대응하기 위해 다양한 전략을 수립 중입니다. 특히, 경쟁 업체들이 전기차에 집중하는 동안 하이브리드차량의 수요가 증가하고 있는 점을 주목하고 있습니다. 현대차는 투싼 HEV와 엘란트라 HEV 모델을 통해 각각 23.5%와 23.7%의 판매 증가를 기록하였으며, 기아는 쏘렌토 HEV와 스포티지 HEV에서 각각 81.9%와 76.4%의 기록적인 판매 증가를 보였습니다. 이러한 성과는 현대차·기아가 하이브리드 시스템을 한층 강화하고 있어 가능해진 결과입니다.\n\n*   ### 4-3. 2025년 판매 목표 및 예상\n\n*   현대차·기아는 2025년 동안 하이브리드 모델의 판매를 더욱 확대할 계획입니다. 특히, '디 올 뉴 싼타페 하이브리드'가 출시되면서 기대되는 판매량 증가가 주효할 것으로 예상됩니다. 이 모델은 1.6리터 직분사 터보차지 하이브리드를 탑재하여 최고출력 232마치를 제공하며, 우수한 연비를 자랑하여 소비자들의 호응을 얻을 것으로 보입니다. 또한, 현대차그룹은 기존의 전동화 생산라인을 활용하여 전기차와 하이브리드차를 동시에 생산할 수 있는 체계를 갖추고 있어, 급증하는 하이브리드 수요에 유연하게 대응할 수 있을 것입니다.\n\n5. 향후 전망 및 결론\n-------------\n\n*   ### 5-1. 미국 자동차 시장의 변화\n\n*   미국 자동차 시장은 지속적으로 변화하고 있으며, 이러한 변화를 주도하는 주요 요소 중 하나는 소비자들의 친환경 차량에 대한 선호도 증가입니다. 특히 전기차와 하이브리드 차량의 수요가 증가하며, 이로 인해 자동차 제조사들은 더욱 혁신적인 기술 개발과 다양한 모델 출시를 통해 시장 점유율을 늘리기 위해 노력하고 있습니다. 현대차·기아는 이러한 시장 변화에 효과적으로 대응하기 위해 전기차와 하이브리드 모델의 생산 및 판매 확대에 집중하고 있습니다.\n\n*   또한, 인플레이션 감축법(IRA)과 같은 정책적 요소도 시장에 큰 영향을 미치고 있습니다. 이 법안은 미국 내 전기차 생산을 증대시키고, 해외 제조업체들이 시장에 진입하기 위한 다양한 인센티브를 제공함으로써 미국 시장의 경쟁 상황을 변화시키고 있습니다. 현대차·기아가 이에 맞춰 미국 내 생산기지를 확장하는 전략은 이러한 시장 변화에 부합하며, 앞으로의 성장은 더욱 가속화될 것으로 예상됩니다.\n\n*   ### 5-2. 현대차·기아의 지속 가능한 성장 전략\n\n*   현대차·기아는 지속 가능한 성장을 위해 친환경 차량 개발에 대한 투자를 지속하고 있으며, 이는 단순히 시장에서의 점유율을 높이는 것을 넘어 환경 보호를 위한 고유한 책임을 다하기 위한 노력이라고 할 수 있습니다. 이들은 특히 하이브리드 차량의 라인업을 확장하고, 향후 출시될 차량에 고성능 하이브리드 시스템을 적용하여 연비 성능을 획기적으로 개선할 계획입니다.\n\n*   하이브리드 모델 시장의 급성장과 관련하여, 현대차·기아는 이미 시장의 요구에 맞춰 다양한 하이브리드 모델을 출시하고 있으며, 이러한 모델들은 각각의 소비자 요구에 부응하고 있습니다. 디 올 뉴 싼타페 하이브리드와 같은 최신 모델의 경우, 성능과 효율성을 동시에 충족시켜 소비자들에게 긍정적인 반응을 얻고 있습니다.\n\n*   ### 5-3. 구매환경 변화에 따른 시장 대응\n\n*   소비자의 구매환경은 급속도로 변화하고 있으며, 이는 자동차 산업에도 직접적인 영향을 미치고 있습니다. 특히 디지털화와 온라인 쇼핑의 확산으로 인해 소비자들은 더 많은 정보를 바탕으로 구매 결정을 내리고 있습니다. 현대차·기아는 이러한 소비자 트렌드에 맞춰 디지털 플랫폼을 구축하고 소비자와의 소통을 강화하는 방안을 마련하고 있습니다.\n\n*   또한, 소비자들이 친환경 차량을 고려할 때 주요하게 생각하는 사항 중 하나는 가격과 비용입니다. 현대차·기아는 전기차 및 하이브리드 차량의 가격 경쟁력을 높이기 위해 생산 공정의 효율성을 극대화하고, 안정적인 부품 공급망을 구축함으로써 최종 소비자에게 보다 합리적인 가격을 제공할 수 있도록 힘쓰고 있습니다. 이러한 노력들은 향후 시장 전망에 긍정적인 영향을 미칠 것으로 보입니다.\n\n결론\n--\n\n*   2025년 미국 자동차 시장에서 현대차·기아는 전기차와 하이브리드 시장에서의 리더십 강화를 위해 지속적인 혁신과 성장 전략을 구사하고 있습니다. 소비자들이 친환경 차량에 대한 선호도를 높이고 있는 가운데, 현대차·기아는 이들의 기대에 부응하기 위해 전기차와 하이브리드 차량의 생산 및 판매 확대에 주력하고 있습니다. 인플레이션 감축법(IRA)과 같은 정책적 변화에도 발 빠르게 대응하며, 현지 생산 비율을 높여 경쟁력을 유지하는 동시에, 소비자들에게 보다 합리적인 가격을 제공할 수 있는 노력을 지속하고 있습니다.\n\n*   또한, 현대차·기아는 하이브리드 모델의 라인업을 확장하며 다양한 소비자 요구에 부합하는 제품 개발에 힘쓰고 있습니다. 더욱이, 디지털화를 통해 소비자와의 소통을 강화하고 정보 접근성을 높여 구매 환경의 변화에 적응하고 있습니다. 이러한 종합적인 접근은 현대차·기아의 지속 가능한 성장 전략의 핵심이며, 향후 자동차 산업의 생태계가 변화하는 과정에서 중요한 역할을 하게 될 것입니다. 결국 현대차·기아의 지속 가능한 성장 전략은 기업의 비전과 맞물려 더 나은 환경을 만들어 가는 데 기여할 것으로 기대됩니다.\n\n용어집\n---\n\n*   인플레이션 감축법(IRA) [법률]: 미국 연방 정부가 환경 친화적인 자동차 구매를 장려하기 위해 시행한 법률로, 소비자에게 세액 공제를 제공합니다.\n\n*   전기차(BEV) [차량 유형]: 물리적인 연료가 아닌 전기를 통해 추진되는 자동차로, 배기가스를 발생하지 않는 친환경 차량입니다.\n\n*   하이브리드(HEV) [차량 유형]: 내연기관과 전기모터를 결합하여 구동되는 자동차로, 연비와 친환경성을 동시에 갖춘 차량입니다.\n\n*   혼류생산 방식 [생산 방식]: 두 가지 이상의 제품을 동일한 생산라인에서 동시에 제조하는 방식으로, 제조 효율성을 높이기 위한 전략입니다.\n\n*   소매 판매 [판매 방식]: 최종 소비자에게 직접 차량을 판매하는 방식을 의미하며, 자동차 업계에서 주로 사용되는 판매 통계입니다.\n\n*   전동화 [기술]: 전기 모터와 배터리를 사용하는 자동차 기술로, 연료 효율성 및 친환경성을 증대시키는 방향으로 발전하고 있습니다.\n\n*   디지털화 [트렌드]: 소비자의 구매 결정 과정에서 정보 접근과 편리함을 제공하기 위해 디지털 플랫폼을 활용하는 과정입니다.\n\n출처 성장이 계속되고 있습니다.\n> \n> 우리의 강력하고 다양한 제품 라인업이 시장 점유율 확대와 고객 수요 충족을\n> \n> 더욱 견인할 것으로 기대합니다.”_\n\n* * *\n\n### **2025년 2월 총판매량 요약**\n\n|  | **Feb-25** | **Feb-24** | **증감률** | **2025년 YTD** | **2024년 8](https://www.irsglobal.com/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0041740462853.jpg)](javascript:popview(\'/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0041740462853.jpg\'))\n\n자료 : MARKLINES\n\n<그림 5> 2024 년 미국의 각 자동차 기업ㆍ브랜드의 HEV 판매 대수\n\n[![Image 9](https://www.irsglobal.com/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0051740462865.jpg)](javascript:popview(\'/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0051740462865.jpg\'))\n\n자료 : MARKLINES\n\nICEV(내연기관차)의 시장 점유율은 여전히 압도적으로 높음\n\n전기차 시장이 계속해서 성장하고 있지만, 2024 년에도 미국 시장의 약 80%를 차지한 것은 ICEV 였다.\n\n<그림 6> 2021~2024 년 미국에서의 ICEV 판매 대수와 비율\n\n(막대 그래프 : 판매 대수(왼쪽 축), 선 그래프 : 비율(오른쪽 축))\n\n[![Image 10](https://www.irsglobal.com/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0061740462881.jpg)](javascript:popview(\'/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/0061740462881.jpg\'))\n\n자료 : MARKLINES\n\n특히 픽업 트럭 및 대형 SUV 등 연비보다 실용성을 중시하는 부문에서의 수요가 뿌리 깊은 상황이다.\n\n대표적인 모델로는 Ford 의 F 시리즈와 GM 의 Silverado 가 있다.\n\n이들은 BEV 버전도 제공하지만, 주력 판매량은 여전히 ICEV 버전이 담당하고 있다.\n\n그 배경에는 미국의 광대한 지리 조건과 라이프스타일에 더하여 일부 지역에서의 충전 인프라 부족이 있다.\n\n[![Image 11](https://www.irsglobal.com/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/po-deu-cha1740462743.jpg)](javascript:popview(\'/img_up/shop_pds/irsglobal1/bbs/rwdboard/add_img/2025/po-deu-cha1740462743.jpg\'))\n\n2025 년의 시장 예측 : 트럼프 정권과 일론 머스크의 영향\n\n2025 년에는 트럼프 정권이 탄생함에 따라 BEV 시장을 둘러싼 환경이 변화할 것으로 예측된다.\n\n구체적으로는 다음과 같은 점이 주목받고 있다.\n\n트럼프 정권의 영향\n\nBEV 에 대한 지원금 폐지 및 규제 완화로 인해 석유 산업이 지원을 얻어 휘발유 가격이 낮아질 전망이다.\n\n이러한 움직임은 ICEV 의 수요를 다시 높일 가능성이 있다. 그리고 HEV 수요도 계속해서 꾸준한 추이를 보인다.\n\n일론 머스크의 영향\n\n앞서 언급한 것처럼 BEV 지원금이 폐지되지만, 테슬라는 이미 강한 가격 경쟁력을 보유하고 있어 지원금 폐지의 영향을 받지 않을 것으로 보인다.\n\n한편 ‘정부효율화부’에 참가함으로써 BEV 시장을 어느 정도 보호하는 시책을 내걸 가능성도 있다.\n\n기본적으로 트럼프는 ‘미국 제일주의’를 강조하는 시책을 추진한다.\n\n과거의 정권하에서도 연료 기준을 완화하고 석유 산업을 지원한 실적이 있으며, 이에 근거하여 국내 산업을 보호하면서도 고용을 창출하는 것을 중시할 가능성이 있다.\n\n또한 미국 내에서 많은 고용을 창출하게 된다면, (물론 경제 안전 보장 등 다양한 리스크를 불식하는 것이 전제되어야 하지만) 예를 들어 BYD 의 공장을 미국에 유치하게 될 가능성도 생길지 모른다. 현시점에서는 중국과의 경제 경쟁을 강조하는 트럼프 정권의 방침을 고려하면 적극적으로 추진할 것이라고 보긴 어렵지만, 완전히 부정할 수만은 없다.\n\n기성 개념에 사로잡히지 않고 ‘미국 제일주의’를 축으로 하여, 지금까지 이야기한 정책이 앞으로도 잇따라 만들어질 것으로 보인다.\n\n국내외 자동차 기업 및 투자가가 (물론 우리도) 시장 동향에 주목하고 있다.\n\n**한편, 1 월 30 일에 미국 운수부가 다음의 각서를 발행하겠다고 발표했다.**\n\n요점은 다음과 같다.\n\n1. ‘자동차에 대한 접근성에 관한 규제 장벽을 제거한다’\n\n2. ‘차량을 선택하는 소비자를 위해 공평한 규제 환경을 확보한다’\n\n3. ‘필요에 따라 휘발유차의 판매를 제한하는 기능을 가지는 주의 배출 규제 면제를 종료한다’\n\n4. ‘다른 기술보다 전기자동차(EV)를 우대하여 개인, 민간 기업, 정부 기관에 EV 구입을 사실상 의무화하는 불공정한 지원금이나 시장의 왜곡을 불러일으키는 정책을 배제할 것을 검토한다’\n\n(대통령령 14154 ‘Unleashing American Energy(미국의 에너지 해방)’ 제 2 항으로부터)\n\n즉 미국은 결코 BEV 를 배제할 생각이 없으며, ‘BEV 와 ICEV 를 어디까지나 같은 조건에서 싸우게 할 것’이라는 뜻이다.\n\n일론 머스크의 입장에서는 테슬라는 이미 충분한 경쟁력을 가지고 있으며, 같은 조건에서도 충분히 싸울 만하다고 판단했다고 볼 수 있다.\n\n(애초에 머스크는 이미 자동차에 힘을 쏟고 있지 않으며, xAI 에 힘을 쏟고 있다는 이야기도 있다)\n\n결론 : 전기화의 발전과 다양한 수요에 대한 대응\n\n2024 년의 미국 시장은 BEV 와 HEV 가 계속해서 성장하는 한편 ICEV 가 여전히 시장의 주력 상품으로 자리하는 현실을 보여주었다.\n\n2025 년에는 정책 전환 및 경쟁 환경의 변화로 인해 각 기업이 다양한 전략을 모색할 필요성이 있다.\n\n특히 BEV 의 성장을 내다보는 장기적인 투자와 단기적으로 수요가 높은 HEV 및 ICEV 에 대응하는 유연한 접근방식이 경쟁 우위성을 확보하는 열쇠가 될 것이다.\n\n소비자의 다양한 니즈를 정확하게 간파하면서 제품 및 서비스를 균형 있게 전개해야 한다.\n\n이것은 쉬운 과제는 아니지만, 시장 및 데이터와의 지속적인 대화를 통해 적절한 판단을 내리는 것이 성공을 위한 필수 요소라 할 수 있다.\n\n**[2025년 국내외 미래자동차 관련 유망 기술, 시장 동향과 사업화 전략]보고서 상세 # 현대자동차 미국 시장 2025년 전망

현대자동차는 2025년 미국 시장에서 강력한 성장세를 나타내며 역대 최대 판매 기록을 목표로 하고 있습니다. 이 전망은 현대자동차가 친환경차와 SUV 라인업을 중심으로 시장 점유율을 확장하고, 다양한 전략적 접근을 통해 경쟁력을 강화하고 있기 때문입니다.

## 미국 시장 내 현대자동차의 판매 성과

2025년 1월, 현대자동차는 미국 시장에서 1월 기준 역대 최대 판매량을 기록하며, 전년 동월 대비 15% 증가한 54,503대를 판매했습니다. 이는 친환경차 판매 증가와 SUV 라인업의 강세에 기인합니다. 하이브리드 차량은 74% 증가, 전기차(EV)는 15% 증가를 보였습니다. 특히, 싼타페 하이브리드와 투싼 하이브리드의 판매가 큰 폭으로 증가했습니다 [출처: novelone 블로그](https://novelone.tistory.com/).

2025년 2월에는 현대자동차의 미국 법인이 전년 동월 대비 3% 증가한 62,032대를 판매하며 역대 가장 높은 2월 판매량을 기록했습니다. 아이오닉 6 모델은 12% 증가를 보였으며, 하이브리드 전기차(HEV) 판매량은 57% 증가했습니다 [출처: Savannah Korean Times](https://savannahkoreatimes.com/).

## 현대자동차의 전략적 접근

현대자동차는 미국 시장에서의 성장을 위해 지속적인 신차 출시와 마케팅 전략을 추진하고 있습니다. 전기차와 하이브리드 차량의 혼류 생산 방식을 도입해 제조 효율성을 높이고 시장 변화에 신속히 대응하고 있습니다. 이로 인해 현대차는 소비자에게 더 많은 선택지를 제공할 수 있게 되었으며, 2024년에는 소매 판매 기준으로 165만 대를 넘는 성과를 예상하고 있습니다 [출처: Goover AI Report](https://seo.goover.ai/report/202504/go-public-report-ko-deb4e471-2938-494b-b19c-6eb5e10b1286-0-0.html).

또한, 현대차는 인플레이션 감축법(IRA) 같은 정책적 변화에도 발 빠르게 대응하여 현지 생산 비율을 늘리며 경쟁력을 유지하고 있습니다. 이러한 전략은 소비자에게 보다 합리적인 가격을 제공하려는 노력의 일환입니다.

## 미래 전망과 도전 과제

현대자동차의 미국 시장 점유율 확장은 다양한 외부 변수에도 직면할 것입니다. 미국의 25% 관세 부과, 원/달러 환율 변동, 주요 원자재 가격 변화 등이 시장 환경에 영향을 미치고 있습니다. 현대차는 이러한 변수들에 대응하여 현지 생산을 확대하고 재고를 효과적으로 활용하여 비용을 절감할 계획입니다 [출처: Goover AI Report](https://seo.goover.ai/report/202506/go-public-report-ko-650e4de3-b7b3-4916-b47d-9293ec9ecd1f-0-0.html).

아울러, 소비자의 친환경차에 대한 선호가 증가함에 따라 현대차는 전기차 및 하이브리드 모델의 경쟁력을 더욱 강화하여 시장에서의 입지를 다질 것입니다. 그러나, 세계 경제의 불확실성과 자동차 시장의 변동성은 여전히 극복해야 할 과제로 남아있습니다.

현대자동차는 친환경차와 SUV 전략을 기반으로 미국 시장에서의 성장을 지속하고 있으며, 다양한 외부 위협에도 대처할 수 있는 강력한 경쟁력을 확보하고 있습니다. 앞으로도 현대차는 친환경차 시장에서의 리더십을 강화하고, 새로운 정책 변화에 적응하며, 지속 가능한 성장을 위해 다양한 전략을 구사할 것입니다. 이러한 노력이 결실을 맺어 현대자동차가 미국 시장에서 계속해서 유리한 위치를 점할 것으로 기대됩니다.
