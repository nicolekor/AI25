```python
# and 게이트

def AND_gate(x1, x2):
  # weights
  w1 = 0.5
  w2 = 0.5
  bias = -0.6

  # forward
  result = x1*w1 + x2*w2 + bias

  # activation
  if result > 0:
    return 1
  else:
    return 0

AND_gate(0, 0), AND_gate(0, 1), AND_gate(1, 0), AND_gate(1, 1)
```

```python
# or 게이트

def OR_gate(x1, x2):
  # weights
  w1 = 0.6
  w2 = 0.6
  bias = -0.5

  # forward
  result = x1*w1 + x2*w2 + bias

  # activation
  if result > 0:
    return 1
  else:
    return 0

OR_gate(0, 0), OR_gate(0, 1), OR_gate(1, 0), OR_gate(1, 1)
```

```python
# NOT 게이트

def NOT_gate(x1):
  # weights
  w1 = -0.6
  bias = 0.5

  # forward
  result = x1*w1 + bias

  # activation
  if result > 0:
    return 1
  else:
    return 0

NOT_gate(0), NOT_gate(1)
```

```python
def step_function(x):
  return 1 if x > 0 else 0

def layer1(x1, x2):
  # weights
  # NAND
  w00 = -0.5
  w01 = -0.5
  b0 = 0.6
  # OR
  w10 = 0.6
  w11 = 0.6
  b1 = -0.5

  # forward
  result0 = x1*w00 + x2*w01 + b0
  result1 = x1*w10 + x2*w11 + b1

  # activation
  result0 = step_function(result0)
  result1 = step_function(result1)

  return result0, result1
```

```python
def layer2(x1, x2):
  # weights
  # AND
  w0 = 0.5
  w1 = 0.5
  bias = -0.7

  # forward
  result = x1*w0 + x2*w1 + bias

  # activation
  return step_function(result)

def XOR_MLP(x1, x2):
  x1, x2 = layer1(x1, x2)
  y = layer2(x1, x2)
  return y

XOR_MLP(0, 0), XOR_MLP(0, 1), XOR_MLP(1, 0), XOR_MLP(1, 1)
```
